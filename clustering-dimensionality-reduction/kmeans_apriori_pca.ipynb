# **PCA**

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_classification
```

```python
# Generate synthetic dataset
X, y = make_classification(n_samples=150, n_features=6, random_state=42)
df1 = pd.DataFrame(X, columns=[f'Feature {i+1}' for i in range(X.shape[1])])
```

```python
# Standardizing the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Applying PCA
pca = PCA(n_components=2)  # Reduce to 2 principal components
X_pca = pca.fit_transform(X_scaled)
```

```python
# Explained variance ratio
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

```python
# Plot the PCA results
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='coolwarm', edgecolors='k', alpha=0.7)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization")
plt.colorbar(label="Class")
plt.show()
```

# **K-MEANS CLUSTERING**

```python
# Elbow Method

from sklearn.cluster import KMeans

data = df3[['Annual Income (k$)', 'Spending Score (1-100)']]
inertias = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(data)
    inertias.append(kmeans.inertia_)

plt.plot(range(1,11), inertias, marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

```python
# Clustering

kmeans = KMeans(n_clusters=5)
kmeans.fit(data)

plt.scatter(data['Annual Income (k$)'], data['Spending Score (1-100)'], c=kmeans.labels_)
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('K-Means Clustering')
plt.show()
```

# **APRIORI**

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
df2 = pd.read_csv('/content/drive/MyDrive/CSE475/Groceries_dataset.csv', encoding="ISO-8859-1")
df2.head()
```

```python
# data has many itemDescription choose any one for check.
df2.itemDescription.value_counts().head(5)
```

```python
df1 = df2[df2.itemDescription == 'yogurt']
```

```python
# some spaces are there in description; need to remove else later operation it will create problem..
df2['itemDescription'] = df2['itemDescription'].str.strip()
```

```python
# convert data in format which is required

basket = pd.pivot_table(data=df2,index='Member_number',columns='itemDescription',values='itemDescription', aggfunc='count',fill_value=0)
```

```python
basket.head()
```

```python
# remove postage item as it is just a seal which almost all transaction contains.
if 'postage' in basket.columns:
    print(basket['postage'].head())
    basket.drop(columns=['postage'], inplace=True)
else:
    print("'postage' not found in basket columns.")
```

```python
# Convert the counts to binary values (0 or 1)
def encode_units(x):
    if x <= 0:
        return 0
    else:
        return 1

basket_sets = basket.applymap(encode_units)
```

```python
# Generate frequent itemsets
frequent_itemsets = apriori(basket_sets, min_support=0.01, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

# Display the top 10 rules
display(rules.head(10))
```

```python
frequent_itemsets = apriori(basket_sets, min_support=0.07, use_colnames=True)
```

```python
#it will generate frequent itemsets
frequent_itemsets
```

```python
# We would apply association rules on frequent itemset.
# here we are setting based on lift and keeping minimum lift as 1

rules_mlxtend = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
rules_mlxtend.head()
```

```python
# rules_mlxtend.rename(columns={'antecedents':'lhs','consequents':'rhs'})

# as based business use case we can sort based on confidance and lift.
rules_mlxtend[ (rules_mlxtend['lift'] >= 4) & (rules_mlxtend['confidence'] >= 0.8) ].head()
```
